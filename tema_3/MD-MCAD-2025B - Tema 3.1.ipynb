Los **árboles de decisión** son modelos de aprendizaje *supervisados* no paramétricos que se utilizan para llevar a cabo tareas, ya sea de clasificación o bien regresiones.

Estos se construyen mediante una serie de *propociones lógicas* secuenciadas, definidas con respecto a un conjunto de datos dado.

```python
Image(filename="resources/decision_tree_explanation.png")
```

Por lo general, un árbol de decisión consta de las siguientes partes:
* 1. **Nodo raíz**
* 2. **Nodos de decisión (o internos)**
* 3. **Nodos hoja (o terminales)**
* 4. **Arcos**
* 5. **Sub-árbol (o ramas)**
